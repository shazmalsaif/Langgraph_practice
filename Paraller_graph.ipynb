```python
from langgraph.graph import StateGraph,START,END
from langchain_huggingface import HuggingFaceEndpoint,ChatHuggingFace
from typing import _TypedDict,Annotated
from pydantic import BaseModel,Field
from dotenv import load_dotenv
```

```python
load_dotenv()  # Load environment variables from .env file
```

```python
class Evulation(BaseModel):
    Feedback: str = Field(description="Returned feedback from the model")
    Score: int = Field(description="Score from 1 to 10")
```

```python
llm = HuggingFaceEndpoint(model="Qwen/Qwen2.5-72B-Instruct",
                          task="text2text-generation"
                          )

model = ChatHuggingFace(llm=llm)
```

```python
Structure_model = model.with_structured_output(
    Evulation,
    method="json_schema"
)
```

```python
Easy ="""Title: The Dual Edge of Synthetic Intelligence: Cognition, Creativity, and Epistemic Authority

The rapid integration of Large Language Models (LLMs) into academic and industrial workflows has fundamentally altered human interactions with information. Rather than acting as mere databases, modern neural networks serve as generative agents capable of synthesizing vast corpuses of text, simulating inductive reasoning, and drafting complex code within seconds. However, as these systems gain widespread adoption, we must critically evaluate their impact on human cognitive autonomy, the authenticity of creative expression, and the nature of epistemic authority.

From a functional perspective, generative AI represents a monumental leap in data synthesis. By utilizing transformer architectures grounded in self-attention mechanisms, models parse context across thousands of tokens simultaneously. This enables them to detect latent pattern in massive datasets that human researchers might overlook. In fields such as bioinformatics and legal research, AI acts as an intellectual force multiplier, dramatically accelerating literature reviews and hypothesis generation. The efficiency gains are undeniable; tasks that once required weeks of painstaking manual cross-referencing can now be executed in minutes.

Yet, this efficiency introduces a subtle paradox regarding depth of knowledge. While an LLM can simulate deep expertise across diverse domains, its output is ultimately derived from probabilistic predictions rather than genuine comprehension. This distinction is critical when evaluating AI as an epistemic authority. When users rely unthinkingly on generated summaries, they risk internalizing "hallucinations"—plausible-sounding statements that lack factual foundation. Furthermore, over-reliance on automated synthesis may erode critical thinking skills, transforming active human learners into passive consumers of pre-digested summaries.

Beyond epistemic concerns, the proliferation of AI-generated content raises profound questions about original thought and linguistic nuance. Human language is not merely a probabilistic sequence of words; it is deeply tied to lived experience, emotional resonance, and cultural context. Although LLMs excel at producing grammatically coherent text, their style often gravitates toward formulaic structures and predictable transitions. When AI tools are overused in creative writing or qualitative research, the subtle idioms and nuanced arguments that define high-caliber human communication risks being diluted into generic uniformity.

In conclusion, synthetic intelligence offers unprecedented utility as an analytical partner, but it cannot serve as a surrogate for human judgment. Maximizing the benefits of this technology requires maintaining a deliberate balance: leveraging LLMs for scalable data processing while preserving rigorous human oversight. Only by maintaining our capacity for independent critical inquiry can we ensure that technological advancement enhances—rather than diminishes—the richness of human intellect."""
```

```python
Prompt = (
    f"You are a helpful assistant. Evaluate the essay below and return ONLY a JSON object with two keys: "
    f"'feedback' and 'score'.\n\nEssay:\n{Easy}"
)
result = Structure_model.invoke(Prompt)
print(result)
```
